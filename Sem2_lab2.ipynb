## АБ тестирование

### > Верно ли, что результаты Надаля на кортах с глиняным покрытием выше, чем на других покрытиях?

```python
import pandas as pd
import numpy as np

from scipy.stats import chi2_contingency
```

```python
total = pd.DataFrame(
    data={"win": [1660, 3658, 5318], "loss": [863, 2715, 3578], 'total': [2523, 6373, 8896]},
    index=['Clay', 'Nonclay', 'total'])

on_serve = pd.DataFrame(
    data={"win": [1090, 171, 1261], "loss": [2747, 446, 3193], 'total': [3837, 617, 4454]},
     index=['Clay', 'Nonclay', 'total'])

on_accept = pd.DataFrame(
    data={"win": [570, 692, 1262], "loss": [911, 2269, 3180], 'total': [1481, 2961, 4442]},
     index=['Clay', 'Nonclay', 'total'])
```

## Проверка гипотез

----
----

```python
chi2, p_value, dof, exp = chi2_contingency(total[['win', 'loss']].iloc[:2].to_numpy(), correction=False)

print(f'p-value: {p_value}')
```

### По данным всех геймов можем сказать, что на глиняном покрытии Надаль стат значимо играет лучше

----

```python
chi2, p_value, dof, exp = chi2_contingency(on_serve[['win', 'loss']].iloc[:2].to_numpy(), correction=False)

print(f'p-value: {p_value}')
```

###  В геймах где Надаль подавал, не можем говорить о значимости покрытия

----

```python
chi2, p_value, dof, exp = chi2_contingency(on_accept[['win', 'loss']].iloc[:2].to_numpy(), correction=False)

print(f'p-value: {p_value}')
```

### В геймах где Надаль принимал, можно утверждать что есть влияние типа покрытия на результат
